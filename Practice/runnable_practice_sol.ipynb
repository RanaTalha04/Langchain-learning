### This is the code for the solution of standard chain problem.

```python
import random
from abc import ABC, abstractmethod
```

```python
# Runnable Class

class Runnable(ABC):
    
    @abstractmethod
    def invoke(input_data):
        pass
```

```python
# LLM Class

class LLM(Runnable):
    
    def __init__(self):
        print("LLM Initiated")

    def invoke(self, prompt):
        
        response_list = [
            'Islamabad is the capital of Pakistan',
            'PSL is a cricket league',
            'AI stands for Artificial Intelligence'
        ]
        
        return {'response': random.choice(response_list)}
        
    def predict(self, prompt):
        
        response_list = [
            'Islamabad is the capital of Pakistan',
            'PSL is a cricket league',
            'AI stands for Artificial Intelligence'
        ]
        print("Warning: This method is depricated and will not be maintained, instead use .invoke() mehtod")
        return {'response': random.choice(response_list)}
```

```python
llm = LLM()
```

```python
# Prompt Template Class

class PromptTemplate:
    
    def __init__(self, template, input_variables):
        
        self.template = template
        self.input_variables = input_variables
        
    def invoke(self, input_dict):
        
        return self.template.format(**input_dict)
    
    def format(self, input_dict):
        print("Warning: This method is depricated and will not be maintained, instead use .invoke() mehtod")        
        return self.template.format(**input_dict)
```

```python
template = PromptTemplate(
    template='Write a {length} lines of poem on {topic}',
    input_variables=['length', 'topic']
)
```

```python
# String Output Parser

class StrOutParser(Runnable):
    def __init__(self):
        pass
    
    def invoke(self, input_data):
        return input_data['response']
```

```python
parser = StrOutParser()
```

```python
# Runnable connectors

class RunnableConnectors(Runnable):
    
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list
        
    def invoke(self, input_data):
        
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
            
        return input_data
```

```python
chain = RunnableConnectors([template, llm, parser])
chain.invoke({'length': 10, 'topic': 'Daffodills'})
```

### Joining different Chains

```python
template1 = PromptTemplate(
    template="Write a joke about {topic}",
    input_variables=['topic']
)

template2 = PromptTemplate(
    template="Explain the following joke. \n{response}",
    input_variables=['response']
)
```

```python
chain1 = RunnableConnectors([template1, llm])
chain2 = RunnableConnectors([template2, llm, parser])
```

```python
final_chain = RunnableConnectors([chain1, chain2])
```

```python
final_chain.invoke({'topic': 'AI'})
```
